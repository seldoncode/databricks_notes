### **1.1 ¿Qué es Databricks y por qué es importante?**

> 💡 **Bienvenido al entorno de datos del futuro**
> Estás leyendo esto dentro de un **Notebook de Databricks**. A diferencia de un documento estático, esto es un entorno vivo donde puedes programar, documentar y visualizar resultados en tiempo real.

#### **¿Qué es Databricks?**

Es la plataforma líder de **Data Intelligence** en la nube. Fue creada por los inventores de **Apache Spark** para unificar todo lo que un equipo de datos necesita en un solo lugar.

Para entenderlo, olvida los servidores y piensa en comida:

| Herramienta | Analogía | Capacidad |
| --- | --- | --- |
| **Excel / Pandas** | Tu cocina de casa | Ideal para platos pequeños (datasets que caben en tu RAM local). |
| **Databricks** | Una cocina industrial | Diseñada para procesar **Petabytes** de datos con eficiencia, orden y muchos cocineros (nodos) trabajando a la vez. |

#### **¿Por qué es el estándar de la industria?**

1. **Escalabilidad**: Usa el motor **Spark** para dividir tareas pesadas entre varias máquinas. Si una tarea tarda 10 horas en tu laptop, Databricks la reparte para que tarde minutos.
2. **Colaboración**: Es como "Google Docs para datos". Varios usuarios pueden editar el mismo código a la vez.
3. **Arquitectura Lakehouse**: Combina lo barato de un "Data Lake" (almacén de archivos) con la velocidad y orden de un "Data Warehouse" (base de datos profesional).

---

#### **💻 Prueba de potencia: Pandas vs Spark**

Ejecuta la siguiente celda. Vamos a comparar una operación tradicional frente al motor distribuido de Databricks.

```python
import pandas as pd
import time
from pyspark.sql.functions import col, sum as spark_sum

print("📊 EXPERIMENTO: Procesando 1 Millón de filas")
print("-" * 40)

# --- ESCENARIO 1: PANDAS (TRADICIONAL) ---
inicio = time.time()
df_pd = pd.DataFrame({'id': range(1000000), 'valor': [i * 2 for i in range(1000000)]})
resultado_pd = df_pd['valor'].sum()
print(f"✅ Pandas finalizado en: {time.time() - inicio:.4f}s")

# --- ESCENARIO 2: SPARK (DATABRICKS) ---
inicio = time.time()
df_spark = spark.range(1000000).withColumn("valor", col("id") * 2)
resultado_spark = df_spark.select(spark_sum("valor")).collect()[0][0]
print(f"🚀 Spark finalizado en:  {time.time() - inicio:.4f}s")

print("-" * 40)
print("💡 NOTA: En la Free Edition la diferencia es sutil, pero con 1.000 Millones de filas, Pandas colapsaría y Spark seguiría volando.")

```

# **Tema 4: El motor de Databricks: Apache Spark**

Si Databricks es la cocina, **Apache Spark** es el equipo de chefs de élite trabajando en paralelo. En este tema entenderás por qué Spark es capaz de procesar volúmenes de datos que harían explotar a un Excel o a un script de Python tradicional.

---

### **4.1 El Cerebro y el Músculo: Arquitectura Spark (Driver & Workers)**

Spark no es un programa que corre en un solo lugar; es un sistema de **computación distribuida**. Para que funcione, el trabajo se divide entre dos roles fundamentales:

1. **El Driver (El Cerebro) 🧠**:
* Es el "Jefe de Cocina".
* Analiza tu código, crea un plan de trabajo y decide cómo repartirlo.
* Mantiene la comunicación con el usuario (tu Notebook).


2. **Los Workers (El Músculo) 💪**:
* Son los "Ayudantes de Cocina".
* Ejecutan las tareas que les manda el Driver.
* Procesan los datos y devuelven los resultados.



> ⚠️ **Nota para el entorno Free**: En la versión *Community*, tienes un solo nodo que hace de Driver y Worker a la vez. Sin embargo, el código que escribes aquí funcionará exactamente igual en un cluster real con **1.000 Workers**.

---

### **4.2 DataFrames: La estructura reina 👑**

En Spark, casi todo lo que hagamos será manipular **DataFrames**.
Un DataFrame es una colección de datos organizada en columnas con nombre. Piensa en él como una tabla de SQL o una hoja de Excel, pero con una diferencia clave: **está distribuido**.

* **Inmutabilidad**: Una vez creado un DataFrame, no se puede cambiar. Si aplicas un filtro, Spark crea un *nuevo* DataFrame con el resultado. Esto parece ineficiente, pero es lo que permite que Spark sea tan seguro y rápido.
* **Escalabilidad**: Un DataFrame puede tener trillones de filas. Spark se encarga de que una parte de esas filas esté en el Worker A y otra parte en el Worker B de forma transparente para ti.

---

### **4.3 El arte de ser "vago": Lazy Evaluation (Transformaciones vs Acciones)**

Spark es extremadamente eficiente porque es **vago** (*Lazy*). No hace nada hasta que no tiene más remedio. Para entender esto, dividimos nuestras órdenes en dos tipos:

#### **A. Transformaciones (El plan de trabajo)**

Son instrucciones como `select()`, `filter()` o `groupBy()`.

* **¿Qué hace Spark?** Solo toma nota en su libreta. No toca los datos todavía.
* **Analogía**: Es como hacer la lista de la compra. Escribir "comprar leche" no trae la leche a tu casa.

#### **B. Acciones (La ejecución)**

Son comandos como `show()`, `count()`, `collect()` o `display()`.

* **¿Qué hace Spark?** Aquí es donde dice: "¡Vale, ahora tengo que trabajar!". Revisa toda la lista de transformaciones, las optimiza (para ir más rápido) y las ejecuta.
* **Analogía**: Es el momento de ir al supermercado y comprar todo lo de la lista de una sola vez.

---

### **4.4 Spark UI: Mirando bajo el capó 🔍**

Cuando ejecutas una celda en Databricks, verás un pequeño enlace debajo que dice **"Spark Jobs"**. Si haces clic, se abre la **Spark UI**.

* **Jobs y Stages**: Aquí puedes ver cómo Spark ha dividido tu código en pequeñas tareas.
* **Optimización**: Puedes ver cuánto tiempo ha tardado cada Worker en hacer su parte.
* **Depuración**: Si una consulta va lenta, la Spark UI te dirá exactamente en qué paso se ha quedado atascado el motor.

> 💡 **Tip**: No hace falta ser un experto en la Spark UI ahora, pero acostúmbrate a abrirla para ver cómo tus órdenes se transforman en tareas paralelas.

---

### **Puntos clave para recordar**

* Spark divide el trabajo entre un **Driver** y múltiples **Workers**.
* Los **DataFrames** son la estructura estándar para manejar datos a escala.
* Gracias a la **Lazy Evaluation**, Spark optimiza tus consultas antes de ejecutarlas.
* Las **Transformaciones** planifican, las **Acciones** ejecutan.

## **Tema 5: Unity Catalog - La Gobernanza Moderna**

Imagina que trabajas en una empresa con cientos de tablas, archivos y notebooks. Sin un sistema de organización, encontrar el dato correcto sería como buscar una aguja en un pajar. **Unity Catalog** es la solución de Databricks para poner orden en el caos.

> 💡 **Unity Catalog es el sistema de gobernanza unificado de Databricks**. Piensa en él como el bibliotecario que sabe exactamente dónde está cada libro (dato), quién lo ha cogido prestado y quién tiene permiso para leerlo.

---

### **5.1 Del caos al orden: Catalog → Schema → Table**

Unity Catalog organiza tus datos en una jerarquía de **tres niveles**, similar a una biblioteca bien estructurada:

| Nivel | Analogía | Descripción | Ejemplo |
|-------|----------|-------------|---------|
| **Catalog** | La biblioteca | El nivel más alto. Normalmente representa un proyecto, departamento o entorno (producción vs pruebas). | `ventas`, `marketing`, `finanzas` |
| **Schema** | La estantería | También llamado "base de datos". Agrupa tablas relacionadas por temática o caso de uso. | `clientes`, `productos`, `transacciones` |
| **Table** | El libro | Los datos reales, organizados en filas y columnas. | `datos_clientes`, `catalogo_productos` |

#### **Ver tu catálogo en acción**

Ejecuta la siguiente celda para explorar la jerarquía en tu entorno:

```sql
%sql
-- Ver todos los catálogos disponibles
SHOW CATALOGS;
```

```sql
%sql
-- Explorar un catálogo específico (normalmente el principal)
SHOW SCHEMAS IN main;
```

```sql
%sql
-- Ver tablas en un schema
SHOW TABLES IN main.default;
```

> ⚠️ **En Free Edition**: Por defecto, trabajarás con el catálogo `main` y el schema `default`. No puedes crear nuevos catálogos (eso es una limitación de la versión gratuita), pero sí puedes crear tablas dentro de `main.default`.

#### **Crear tu primera tabla organizada**

```python
%python
# Crear un DataFrame de ejemplo
datos_estudiantes = spark.createDataFrame([
    (1, "Ana García", 23, "Matemáticas"),
    (2, "Carlos López", 24, "Física"),
    (3, "Elena Martínez", 22, "Ingeniería")
], ["id", "nombre", "edad", "carrera"])

# Guardar como tabla en Unity Catalog (catalog.schema.tabla)
datos_estudiantes.write.saveAsTable("main.default.estudiantes")

print("✅ Tabla creada en main.default.estudiantes")
```

```sql
%sql
-- Verificar que la tabla existe y está bien organizada
SELECT * FROM main.default.estudiantes;
```

### **5.2 Permisos y lineage de datos**

La gobernanza no es solo organización física; también es control de acceso y trazabilidad.

#### **Permisos (quién puede hacer qué)**

En un entorno empresarial, no todo el mundo debe tener acceso a todos los datos. Unity Catalog gestiona permisos a nivel de:
- **Catalog**: Acceso completo a un proyecto
- **Schema**: Acceso a un área funcional
- **Table**: Acceso a tablas específicas
- **Row/Column**: Acceso a filas o columnas concretas (control fino)

**Comandos de permisos (solo visualización en Free Edition):**

```sql
%sql
-- Ver permisos de una tabla (en Free Edition solo verás tu propia tabla)
SHOW GRANTS ON TABLE main.default.estudiantes;
```

> 💡 **Tip**: En la Free Edition no podrás gestionar permisos complejos, pero es crucial que entiendas este concepto. En una empresa real, tú como ingeniero de datos asignarías permisos como:
> ```sql
> GRANT SELECT ON TABLE ventas.clientes.datos TO analyst@empresa.com;
> ```

#### **Lineage (de dónde vienen los datos)**

El **lineage** o linaje de datos es la capacidad de trazar el origen y las transformaciones de un dato. Es como el árbol genealógico de tus datos.

**¿Para qué sirve?**
- **Auditoría**: Saber quién creó una tabla y con qué datos
- **Depuración**: Si un informe está mal, puedes ver qué transformaciones se aplicaron
- **Impacto**: Antes de eliminar una tabla, sabes qué otros procesos dependen de ella

**Ver lineage en acción:**

1. Ejecuta esta celda para crear una tabla derivada:
```sql
%sql
-- Crear una vista que depende de la tabla original
CREATE OR REPLACE VIEW main.default.estudiantes_mayores AS
SELECT nombre, carrera 
FROM main.default.estudiantes 
WHERE edad > 22;
```

2. **Para ver el lineage** (en interfaz gráfica):
   - Ve al **Catalog** en la barra lateral izquierda
   - Navega a `main.default.estudiantes`
   - Haz clic en la pestaña **"Lineage"**
   - Verás un gráfico que muestra qué tablas dependen de esta

> 💡 **Tip profesional**: El lineage es automático en Databricks. Cada vez que lees una tabla para crear otra, Unity Catalog registra esa relación.

### **5.3 Volumes vs DBFS (cuándo usar cada uno)**

Este es uno de los conceptos más confusos para principiantes. Vamos a aclararlo de una vez.

#### **¿Qué es DBFS? (El pasado)**

**DBFS (Databricks File System)** era la forma antigua de almacenar archivos. Piensa en él como el "disco duro" de Databricks.

**Características:**
- Ruta típica: `/mnt/datos/archivo.csv`
- **Problema**: Cualquier usuario con acceso al workspace podía leer/escribir en DBFS (¡caos de seguridad!)
- **Legacy**: Sigue existiendo por compatibilidad, pero ya no es recomendable

#### **¿Qué son los Volumes? (El presente)**

**Volumes** son la evolución moderna dentro de Unity Catalog. Son directorios de archivos gobernados con los mismos permisos que las tablas.

**Características:**
- Ruta típica: `/Volumes/main/default/mis_datos/archivo.csv`
- **Ventaja**: Los permisos de Unity Catalog también aplican a archivos
- **Organización**: Los volumes viven dentro de Catalog → Schema

#### **Comparación directa**

| Aspecto | DBFS (Legacy) | Volumes (Moderno) |
|---------|---------------|-------------------|
| **Gobernanza** | ❌ Ninguna (acceso global) | ✅ Controlada por Unity Catalog |
| **Permisos** | ❌ No se pueden asignar | ✅ GRANT/REVOKE como en tablas |
| **Visibilidad** | ❌ Fuera de Catalog | ✅ Visible en la interfaz de Catalog |
| **Uso recomendado** | ⚠️ Solo sistemas heredados | ✅ Todos los proyectos nuevos |

#### **Crear y usar un Volume (Demostración)**

En Free Edition no puedes crear Volumes, pero puedes simular su estructura:


| Si necesitas... | Usa... | Por qué |
|-----------------|--------|---------|

| **Compatibilidad con código antiguo** | ⚠️ DBFS | Solo mientras migras a Volumes |

#### **Puntos clave para recordar**

### **6.1 Unity Catalog Volumes: El nuevo estándar**

Un **Volume** es un directorio de archivos gestionado por Unity Catalog. Piensa en él como una carpeta compartida en la nube, pero con superpoderes:

> 💡 **Analogía**: Si Unity Catalog es la biblioteca, los Volumes son las estanterías donde apilamos los archivos antes de encuadernarlos como libros (tablas).

### **6.4 Tablas en Unity Catalog (managed vs external)**

Cuando guardamos datos en Unity Catalog, tenemos dos opciones fundamentales. Esta decisión afecta a quién controla los archivos físicamente.

#### **Managed Tables (Tablas Gestionadas)**

| Aspecto | Descripción |
|---------|-------------|
| **Quién gestiona los archivos** | Databricks |
| **Dónde se guardan** | Ubicación interna del Catalog |
| **Al eliminar la tabla** | Se borran LOS DATOS y los metadatos |
| **Caso de uso** | Tablas intermedias, resultados de ETL, datos efímeros |

#### **External Tables (Tablas Externas)**

| Aspecto | Descripción |
|---------|-------------|
| **Quién gestiona los archivos** | Tú (en tu cloud storage) |
| **Dónde se guardan** | Ubicación externa (S3, ADLS, GCS) que tú especificas |
| **Al eliminar la tabla** | Solo se borran los metadatos, LOS DATOS PERMANECEN |
| **Caso de uso** | Datos críticos, tablas compartidas con otros sistemas |

#### **Crear tabla managed (simulación)**

```python
%python
# Esto crea una MANAGED table (los datos los gestiona Databricks)
df_ventas.write.saveAsTable("main.default.ventas_managed")

print("✅ Tabla managed creada: main.default.ventas_managed")
print("📁 Ubicación: Gestionada internamente por Databricks")
```

#### **Crear tabla external (simulación con DBFS)**

En producción, una external table apunta a tu cloud storage. Aquí simulamos apuntando a DBFS:

```python
%python
# Guardar los datos primero en nuestra "ubicación externa simulada"
ruta_externa_simulada = "dbfs:/tmp/external_data/ventas/"

USING CSV

  header = 'true',

)
""")

print("✅ Tabla external creada: main.default.ventas_external")
print(f"📁 Apunta a: {ruta_externa_simulada}")
```

### **6.5 DBFS legacy: cuándo aún lo verás**

Aunque Volumes son el estándar actual, seguirás viendo DBFS en muchos lugares:

#### **¿Dónde sobrevive DBFS?**

1. **Documentación antigua**: Tutoriales de 2020-2022 usan DBFS
2. **Código heredado**: Proyectos que empezaron antes de Unity Catalog
3. **Áreas temporales**: El directorio `/tmp/` sigue siendo útil para archivos efímeros

#### **Cuándo está bien usar DBFS hoy**

| Tablas de producción | ❌ No | Tablas managed/external en Catalog |
| Prácticas de aprendizaje (Free Edition) | ⚠️ Sí, forzado | Simular estructura de Volumes |

## **Tema 8: Delta Lake: El Corazón del Lakehouse**

Has aprendido a cargar datos, limpiarlos y transformarlos con PySpark. Pero hay un problema: cuando guardas un archivo CSV o Parquet, no tienes control sobre lo que pasa si dos personas escriben a la vez, ni puedes recuperar versiones anteriores si alguien comete un error.

**Delta Lake** es la solución a todos estos problemas. Es la capa que convierte tu data lake en un lakehouse, dándole superpoderes de base de datos a tus archivos.

> 💡 **Delta Lake es formato de almacenamiento + transacciones ACID + versionado + optimizaciones**. Todo en uno.

---

### **8.1 ¿Por qué Delta Lake? (Transacciones ACID)**

Para entender Delta Lake, primero hay que entender los problemas del mundo real:

#### **El problema: Sin ACID**

Imagina que dos procesos intentan escribir en la misma tabla a la vez:
- **Proceso A**: Actualiza 1.000 filas
- **Proceso B**: Lee la tabla para un informe justo en medio de la actualización

**Sin Delta Lake**: El Proceso B vería datos inconsistentes (mitad actualizados, mitad no). Peor aún, si el Proceso A falla a mitad, te quedan datos corruptos.

**Con Delta Lake**: Las escrituras son **transaccionales**. O se hace todo, o no se hace nada. Y las lecturas siempre ven una versión consistente.

#### **Las 4 propiedades ACID**

| Propiedad | Significado | En Delta Lake |
|-----------|-------------|---------------|
| **A**tomicity | Una transacción se hace completa o no se hace | Las escrituras son todo o nada |
| **C**onsistency | Los datos siempre cumplen reglas definidas | Esquemas validados |
| **I**solation | Transacciones simultáneas no interfieren | Lecturas ven estado consistente |
| **D**urability | Una vez guardado, no se pierde | Los datos persisten en cloud storage |

### **8.2 Time Travel: Viajando al pasado de tus datos**

El **Time Travel** es una de las características más impresionantes de Delta Lake. Puedes consultar cómo eran tus datos en cualquier momento del pasado.

#### **Versionado automático**

### **8.3 Evolución de esquemas y optimización (Z-ORDER)**

#### **Evolución de esquemas (Schema Evolution)**

En el mundo real, los datos cambian. Nuevas columnas aparecen, tipos de datos se modifican. Delta Lake maneja esto de forma elegante.

**Por defecto, Delta es estricto:**

```python
%python
# Intentar insertar datos con columna extra (fallará)
nuevos_datos_con_extra = spark.createDataFrame([
    Row(id=8, producto="Tablet", cantidad=2, precio=300.00, fecha=date(2024, 1, 19), descuento=10),  # Columna extra
])

try:
    nuevos_datos_con_extra.write.format("delta").mode("append").saveAsTable("main.default.ventas_delta")
except Exception as e:
    print("❌ Error esperado: No coincide el esquema")
    print(f"Error: {str(e)[:200]}...")
```

**Solución 1: Permitir evolución explícitamente**

```python
%python
# Permitir evolución de esquema
nuevos_datos_con_extra.write.format("delta") \
    .option("mergeSchema", "true") \
    .mode("append") \
    .saveAsTable("main.default.ventas_delta")
    
print("✅ Esquema evolucionado: nueva columna 'descuento' añadida")
```

```sql
%sql
-- Verificar que la nueva columna existe
DESCRIBE main.default.ventas_delta;
```

```sql
%sql
-- Ver los datos (descuento será null para registros antiguos)
SELECT * FROM main.default.ventas_delta ORDER BY id;
```

**Solución 2: Sobrescribir esquema completamente**

```python
%python
# Sobrescribir tabla con nuevo esquema (¡CUIDADO! Borra datos)
datos_nuevo_formato = spark.createDataFrame([
    Row(id=9, nombre_producto="Smartwatch", precio=199.99, stock=50),  # Esquema completamente diferente
])

datos_nuevo_formato.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("main.default.ventas_delta")
```


> ⚠️ **Advertencia**: `overwriteSchema` es destructivo. Los datos antiguos se pierden (aunque puedes recuperarlos con Time Travel).

#### **Optimización: Z-ORDER**

Cuando las tablas crecen (millones o miles de millones de filas), las consultas pueden volverse lentas. **Z-ORDER** es una técnica de optimización que reorganiza los datos para que las consultas comunes sean más rápidas.

**El problema: Escaneo completo de tabla**

```sql
%sql
-- Sin optimización, esta consulta escanea TODOS los archivos
SELECT * FROM main.default.ventas_delta WHERE producto = 'Portátil';
```

**Crear datos grandes para probar optimización**

```python
%python
# Crear dataset más grande para sentir la optimización
from pyspark.sql.functions import rand, round, lit, concat
import random

# Generar 100,000 filas de ejemplo
df_grande = spark.range(100000) \
    .withColumn("producto", concat(lit("Producto_"), (rand() * 100).cast("int"))) \
    .withColumn("categoria", 
                when(rand() < 0.33, lit("Electrónica"))
                .when(rand() < 0.66, lit("Accesorios"))
                .otherwise(lit("Audio"))) \
    .withColumn("precio", round(rand() * 500 + 10, 2)) \
    .withColumn("cantidad", (rand() * 10 + 1).cast("int")) \
    .withColumn("fecha", lit("2024-01-20"))

df_grande.write.format("delta").mode("overwrite").saveAsTable("main.default.ventas_grandes")
print("✅ Tabla grande creada con 100,000 filas")
```

**Optimizar con Z-ORDER**

```sql
%sql
-- Optimizar la tabla y aplicar Z-ORDER por columna de filtrado común
OPTIMIZE main.default.ventas_grandes
ZORDER BY (categoria, producto);
```

**¿Qué acaba de pasar?**

- `OPTIMIZE` compacta archivos pequeños en archivos más grandes (mejor rendimiento)
- `ZORDER BY` reorganiza los datos para que filas con valores similares estén cerca físicamente
- Cuando consultes por `categoria` o `producto`, Spark leerá muchos menos archivos

**Comparar rendimiento antes/después**

```sql
%sql
-- Consulta antes de OPTIMIZE (ejecuta y mira el tiempo)
SELECT categoria, COUNT(*) as total, AVG(precio) as precio_medio
FROM main.default.ventas_grandes
WHERE categoria = 'Electrónica'
GROUP BY categoria;
```

```sql
%sql
-- Mismo query después de OPTIMIZE (debería ser más rápido)
SELECT categoria, COUNT(*) as total, AVG(precio) as precio_medio
FROM main.default.ventas_grandes
WHERE categoria = 'Electrónica'
GROUP BY categoria;
```

> 💡 **Tip**: Usa Z-ORDER en columnas que aparecen frecuentemente en `WHERE`, `JOIN` o `GROUP BY`. No abuses: Z-ORDER en demasiadas columnas diluye el beneficio.

#### **Otras optimizaciones importantes**

**Vacuum: Limpiar archivos viejos**

Con Time Travel, Delta guarda todas las versiones. Con el tiempo, esto ocupa mucho espacio.

```sql
%sql
-- Ver archivos no necesarios para retención (por defecto 7 días)
DESCRIBE HISTORY main.default.ventas_grandes;
```

```sql
%sql
-- Limpiar archivos más antiguos de 7 días (¡CUIDADO! pierdes Time Travel más allá)
VACUUM main.default.ventas_grandes RETAIN 168 HOURS;
```

```python
%python
# Verificar archivos después del VACUUM
display(dbutils.fs.ls("dbfs:/user/hive/warehouse/main_default.db/ventas_grandes"))
```

> ⚠️ **Advertencia**: `VACUUM` elimina archivos físicamente. Después de hacer VACUUM, no podrás viajar en el tiempo a versiones anteriores a la retención especificada.

**Liquid clustering (novedad)**

En versiones recientes de Databricks, existe **Liquid Clustering** que reemplaza a Z-ORDER con una optimización continua y automática:

```sql
%sql
-- Crear tabla con liquid clustering (Databricks Runtime 13.3+)
CREATE OR REPLACE TABLE main.default.ventas_cluster
USING delta
CLUSTER BY (categoria, producto)
AS SELECT * FROM main.default.ventas_grandes;
```

Pero para la mayoría de casos, Z-ORDER sigue siendo excelente.

### **9.3 Funciones de ventana (Window Functions)**

Las funciones de ventana son la navaja suiza del análisis SQL. Permiten hacer cálculos a través de filas relacionadas sin perder el detalle de cada fila.

#### **Preparación: Entendiendo el problema**

Sin ventanas, para calcular el ranking de productos por categoría, tendrías que hacer subconsultas complejas. Con ventanas, es elegante y eficiente.

```sql
%sql
-- Ranking de productos por ingresos dentro de cada categoría
SELECT 
    nombre_producto,
    categoria,
    ROUND(SUM(importe_total), 2) as ingresos_totales,
    RANK() OVER (PARTITION BY categoria ORDER BY SUM(importe_total) DESC) as rank_categoria,
    DENSE_RANK() OVER (PARTITION BY categoria ORDER BY SUM(importe_total) DESC) as dense_rank_categoria,
    ROW_NUMBER() OVER (PARTITION BY categoria ORDER BY SUM(importe_total) DESC) as row_num
FROM main.default.ventas
GROUP BY nombre_producto, categoria
ORDER BY categoria, rank_categoria;
```

#### **Tipos de funciones de ventana**

**1. Funciones de ranking**

```sql
%sql
-- Comparar diferentes rankings
SELECT 
    vendedor_id,
    nombre_vendedor,
    region,
    ROUND(SUM(importe_total), 2) as ventas_totales,
    RANK() OVER (ORDER BY SUM(importe_total) DESC) as rank_global,
    DENSE_RANK() OVER (ORDER BY SUM(importe_total) DESC) as dense_rank,
    PERCENT_RANK() OVER (ORDER BY SUM(importe_total) DESC) as percent_rank,
    NTILE(4) OVER (ORDER BY SUM(importe_total) DESC) as quartil
FROM main.default.ventas
GROUP BY vendedor_id, nombre_vendedor, region
ORDER BY ventas_totales DESC;
```

**2. Funciones de desplazamiento (LAG/LEAD)**

Para comparar con filas anteriores o posteriores:

```sql
%sql
-- Ventas diarias y comparación con día anterior
WITH ventas_diarias AS (
    SELECT 
        fecha,
        ROUND(SUM(importe_total), 2) as ventas_dia
    FROM main.default.ventas
    GROUP BY fecha
)
SELECT 
    fecha,
    ventas_dia,
    LAG(ventas_dia, 1) OVER (ORDER BY fecha) as ventas_dia_anterior,
    LEAD(ventas_dia, 1) OVER (ORDER BY fecha) as ventas_dia_siguiente,
    ROUND(ventas_dia - LAG(ventas_dia, 1) OVER (ORDER BY fecha), 2) as diferencia_dia_anterior,
    ROUND((ventas_dia / LAG(ventas_dia, 1) OVER (ORDER BY fecha) - 1) * 100, 2) as crecimiento_porcentual
FROM ventas_diarias
ORDER BY fecha;
```

**3. Funciones de agregación con ventana**

Para totales acumulados o medias móviles:

```sql
%sql
-- Ventas acumuladas por vendedor a lo largo del tiempo
SELECT 
    fecha,
    nombre_vendedor,
    importe_total,
    SUM(importe_total) OVER (PARTITION BY nombre_vendedor ORDER BY fecha) as acumulado_vendedor,
    AVG(importe_total) OVER (PARTITION BY nombre_vendedor ORDER BY fecha ROWS BETWEEN 2 PRECEDING AND CURRENT ROW) as media_movil_3_ventas,
    SUM(importe_total) OVER (ORDER BY fecha) as acumulado_global
FROM main.default.ventas
WHERE nombre_vendedor IN ('Ana García', 'Carlos López')
ORDER BY nombre_vendedor, fecha;
```

#### **Frame specification: El secreto de las ventanas**

La cláusula `ROWS BETWEEN` define qué filas incluir en el cálculo:

```sql
%sql
-- Diferentes ventanas para el mismo cálculo
SELECT 
    fecha,
    importe_total,
    -- Desde el principio hasta ahora
    SUM(importe_total) OVER (ORDER BY fecha ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) as acumulado_total,
    -- Últimas 3 ventas (incluyendo actual)
    AVG(importe_total) OVER (ORDER BY fecha ROWS BETWEEN 2 PRECEDING AND CURRENT ROW) as media_3_ventas,
    -- Ventana centrada: 1 anterior, actual, 1 siguiente
    AVG(importe_total) OVER (ORDER BY fecha ROWS BETWEEN 1 PRECEDING AND 1 FOLLOWING) as media_centrada,
    -- Solo filas anteriores (excluyendo actual)
    MAX(importe_total) OVER (ORDER BY fecha ROWS BETWEEN UNBOUNDED PRECEDING AND 1 PRECEDING) as maximo_hasta_antes
FROM main.default.ventas
ORDER BY fecha
LIMIT 20;
```

#### **Casos de uso reales**

**Caso 1: Detectar mejores clientes por categoría**

```sql
%sql
-- Top 3 productos por categoría
WITH ranking_productos AS (
    SELECT 
        categoria,
        nombre_producto,
        ROUND(SUM(importe_total), 2) as ingresos,
        ROW_NUMBER() OVER (PARTITION BY categoria ORDER BY SUM(importe_total) DESC) as posicion
    FROM main.default.ventas
    GROUP BY categoria, nombre_producto
)
SELECT * FROM ranking_productos
WHERE posicion <= 3
ORDER BY categoria, posicion;
```

**Caso 2: Comparar vendedor con promedio de su región**

```sql
%sql
-- Rendimiento relativo por región
WITH ventas_vendedor AS (
    SELECT 
        region,
        nombre_vendedor,
        ROUND(SUM(importe_total), 2) as ventas_vendedor
    FROM main.default.ventas
    GROUP BY region, nombre_vendedor
),
stats_region AS (
    SELECT 
        region,
        AVG(ventas_vendedor) as media_region,
        STDDEV(ventas_vendedor) as desviacion_region
    FROM ventas_vendedor
    GROUP BY region
)
SELECT 
    v.region,
    v.nombre_vendedor,
    v.ventas_vendedor,
    ROUND(s.media_region, 2) as media_region,
    ROUND(v.ventas_vendedor - s.media_region, 2) as diferencia_media,
    ROUND((v.ventas_vendedor - s.media_region) / NULLIF(s.desviacion_region, 0), 2) as z_score
FROM ventas_vendedor v
JOIN stats_region s ON v.region = s.region
ORDER BY v.region, v.ventas_vendedor DESC;
```

**Caso 3: Detectar valores atípicos (outliers)**

```sql
%sql
-- Identificar ventas inusualmente altas para cada vendedor
WITH stats_vendedor AS (
    SELECT 
        nombre_vendedor,
        importe_total,
        AVG(importe_total) OVER (PARTITION BY nombre_vendedor) as media_vendedor,
        STDDEV(importe_total) OVER (PARTITION BY nombre_vendedor) as stddev_vendedor,
        COUNT(*) OVER (PARTITION BY nombre_vendedor) as num_ventas
    FROM main.default.ventas
)
SELECT 
    nombre_vendedor,
    importe_total,
    ROUND(media_vendedor, 2) as media,
    ROUND(stddev_vendedor, 2) as desviacion,
    ROUND((importe_total - media_vendedor) / stddev_vendedor, 2) as desviaciones_media,
    CASE 
        WHEN ABS(importe_total - media_vendedor) > 2 * stddev_vendedor THEN '🚨 OUTLIER'
        ELSE 'Normal'
    END as tipo
FROM stats_vendedor
WHERE num_ventas > 5  -- Suficientes datos para estadística
ORDER BY desviaciones_media DESC;
```

#### **Comparación: Sin ventanas vs Con ventanas**

**Sin ventanas (complicado):**

```sql
%sql
-- Subconsulta correlacionada (ineficiente)
SELECT 
    v1.nombre_producto,
    v1.categoria,
    ROUND(SUM(v1.importe_total), 2) as ingresos,
    (
        SELECT COUNT(DISTINCT v2.nombre_producto) + 1
        FROM main.default.ventas v2
        WHERE v2.categoria = v1.categoria
        GROUP BY v2.categoria
        HAVING SUM(v2.importe_total) > SUM(v1.importe_total)
    ) as ranking
FROM main.default.ventas v1
GROUP BY v1.nombre_producto, v1.categoria
ORDER BY categoria, ranking;
```

**Con ventanas (elegante):**

```sql
%sql
-- Mismo resultado, 10x más legible, 100x más eficiente
SELECT 
    nombre_producto,
    categoria,
    ROUND(SUM(importe_total), 2) as ingresos,
    RANK() OVER (PARTITION BY categoria ORDER BY SUM(importe_total) DESC) as ranking
FROM main.default.ventas
GROUP BY nombre_producto, categoria
ORDER BY categoria, ranking;
```

> 💡 **Tip**: Siempre que necesites comparar filas con otras filas del mismo grupo, piensa en ventanas. Son más rápidas y legibles que las subconsultas.

```python
from datetime import date, timedelta
import random

# Generar datos de ventas por producto, región y tiempo
categorias = ["Electrónica", "Accesorios", "Audio", "Almacenamiento", "Oficina"]
regiones = ["Norte", "Sur", "Centro", "Este", "Oeste"]
productos_por_categoria = {
    "Electrónica": ["Portátil", "Monitor", "Tablet", "Webcam"],
    "Accesorios": ["Teclado", "Ratón", "Funda", "Cargador"],
    "Audio": ["Auriculares", "Altavoces", "Micrófono"],
    "Almacenamiento": ["SSD", "HDD Externo", "USB"],
    "Oficina": ["Impresora", "Scanner", "Destructora"]
}

ventas = []
fecha_inicio = date(2024, 1, 1)

    producto = random.choice(productos_por_categoria[categoria])

    ventas.append((

        fecha,
        producto,
        categoria,
        region,
        cantidad,
        precio_unitario,
        importe

    "cantidad", "precio_unitario", "importe"
])
```

### **11.2 Encadenamiento de notebooks**

Un job puede tener múltiples tareas que se ejecutan en secuencia o en paralelo. Esto se llama **DAG (Directed Acyclic Graph)** de tareas.

#### **Escenario: Pipeline completo nocturno**

Imagina este flujo:
1. **Extraer** datos de fuentes externas
2. **Transformar** y limpiar
3. **Calcular** métricas de negocio
4. **Alertar** si hay anomalías
5. **Refrescar** dashboards

#### **Configuración de tareas encadenadas**

Desde la interfaz de creación/edición de jobs:

**Tarea 1: Extract**
- Nombre: `01_Extract`
- Notebook: `etl/extract_datos_fuente`
- Dependencias: Ninguna (primera tarea)

**Tarea 2: Transform**
- Nombre: `02_Transform`
- Notebook: `etl/clean_and_transform`
- Dependencias: `01_Extract` (solo se ejecuta cuando la primera termina bien)

**Tarea 3: Calculate Metrics**
- Nombre: `03_Calculate_Metrics`
- Notebook: `analytics/calcular_metricas`
- Dependencias: `02_Transform`

**Tarea 4: Alerts** (puede ejecutarse en paralelo)
- Nombre: `04_Alerts`
- Notebook: `monitoring/alertas`
- Dependencias: `02_Transform` (también depende de Transform)

**Tarea 5: Refresh Dashboard**
- Nombre: `05_Refresh_Dashboard`
- Tipo: SQL (en lugar de notebook)
- Consulta: `REFRESH TABLE main.default.dashboard_ventas`
- Dependencias: `03_Calculate_Metrics`

#### **Visualización del DAG**

En la interfaz de Workflows, verías algo así:

```
01_Extract
    ↓
02_Transform
    ├──→ 03_Calculate_Metrics → 05_Refresh_Dashboard
    └──→ 04_Alerts
```

### **11.3 Notificaciones y control de errores**

Un job que se ejecuta solo no sirve de nada si no sabemos qué ha pasado. Las notificaciones son cruciales.

#### **Configurar notificaciones en el job**

En la configuración del job, hay una sección **"Notifications"**:

- **On start**: Cuando el job empieza (opcional)
- **On success**: Cuando termina bien
- **On failure**: Cuando falla (¡IMPORTANTE!)
- **On duration warning**: Si tarda más de lo esperado

**Destinos de notificaciones:**
- **Email**: Lista de direcciones de email
- **Slack**: Canal de Slack (requiere configuración)
- **Webhook**: Llamada a una URL (para integraciones custom)

#### **Manejo de errores dentro del notebook**

Tu código debe ser robusto ante fallos:

```python
%python
# Notebook: proceso_con_errores.py

from pyspark.sql.utils import AnalysisException
import traceback

try:
    print("🔄 Iniciando proceso...")
    
    # Intentar leer tabla que podría no existir
    df = spark.sql("SELECT * FROM main.default.tabla_que_no_existe")
    
    # Procesar...
    df.write.saveAsTable("main.default.resultado")
    
    print("✅ Proceso completado")
    
except AnalysisException as e:
    print(f"❌ Error de tabla: {str(e)}")
    # Enviar alerta específica para este error
    dbutils.notebook.exit(f"ERROR_TABLA: {str(e)}")
    
except Exception as e:
    print(f"❌ Error inesperado: {str(e)}")
    traceback.print_exc()
    # Re-lanzar la excepción para que el job marque fallo
    raise
```

```python
%python
from pyspark.sql import Row
import random
from datetime import datetime, timedelta

# Generar datos de clientes para predicción de compra
clientes = []
for i in range(1, 1001):
    # Características del cliente
    edad = random.randint(18, 70)
    antiguedad = random.randint(0, 120)  # meses
    ingresos = random.randint(20000, 120000)
    visitas_semana = random.randint(0, 10)
    gasto_previo = random.randint(0, 5000)
    
    # Variable objetivo: ¿compró en el último mes? (1= Sí, 0= No)
    # Regla subyacente (simulada)
    probabilidad = 0.1 + (visitas_semana * 0.05) + (gasto_previo / 10000) - (edad / 200)
    probabilidad = max(0, min(1, probabilidad))
    compra = 1 if random.random() < probabilidad else 0
    
    clientes.append((
        i, edad, antiguedad, ingresos, visitas_semana, gasto_previo, compra
    ))

df_clientes = spark.createDataFrame(clientes, [
    "cliente_id", "edad", "antiguedad_meses", "ingresos_anuales", 
    "visitas_semana", "gasto_previo_total", "compra_reciente"
])

# Guardar tabla
df_clientes.write.mode("overwrite").saveAsTable("main.default.clientes_ml")

print("✅ Tabla 'clientes_ml' creada con 1000 registros")
print(f"📊 Distribución objetivo: {df_clientes.groupBy('compra_reciente').count().collect()}")
```

```python
# Seleccionar features y target
feature_columns = ["edad", "antiguedad_meses", "ingresos_anuales", "visitas_semana", "gasto_previo_total"]
target_column = "compra_reciente"

# Dividir en train/test
train_df, test_df = df_clientes.randomSplit([0.8, 0.2], seed=42)

# Assembler: combina features en un vector
assembler = VectorAssembler(inputCols=feature_columns, outputCol="features")

# Modelo
rf = RandomForestClassifier(
    labelCol=target_column,
    featuresCol="features",
    numTrees=100,
    maxDepth=5,
    seed=42
)

# Pipeline
pipeline = Pipeline(stages=[assembler, rf])

# INICIAMOS EL SEGUIMIENTO CON MLFLOW
with mlflow.start_run(run_name="RandomForest_experimento_1"):
    
    # Registrar parámetros manualmente
    mlflow.log_param("model_type", "RandomForest")
    mlflow.log_param("num_trees", 100)
    mlflow.log_param("max_depth", 5)
    mlflow.log_param("train_test_split", "80/20")
    
    # Entrenar
    model = pipeline.fit(train_df)
    
    # Predecir
    predictions = model.transform(test_df)
    
    # Evaluar
    evaluator = BinaryClassificationEvaluator(
        labelCol=target_column,
        rawPredictionCol="rawPrediction",
        metricName="areaUnderROC"
    )
    
    auc = evaluator.evaluate(predictions)
    print(f"✅ AUC: {auc:.4f}")
    
    # Registrar métrica
    mlflow.log_metric("auc", auc)
    
    # También podemos registrar métricas adicionales
    from pyspark.ml.evaluation import MulticlassClassificationEvaluator
    
    evaluator_accuracy = MulticlassClassificationEvaluator(
        labelCol=target_column,
        predictionCol="prediction",
        metricName="accuracy"
    )
    accuracy = evaluator_accuracy.evaluate(predictions)
    mlflow.log_metric("accuracy", accuracy)
    
    print(f"✅ Accuracy: {accuracy:.4f}")
    
    # Registrar el modelo (¡importante!)
    mlflow.spark.log_model(model, "random_forest_model")
    
    # Podemos incluso registrar artefactos
    with open("/tmp/feature_importance.txt", "w") as f:
        f.write(f"Features usados: {feature_columns}")
    mlflow.log_artifact("/tmp/feature_importance.txt")
    
print("✅ Experimento registrado en MLflow")
```

#### **Ejemplo 1: Separar lógica en módulos Python**

**Archivo: `src/etl/transform.py`**

```python
# Este es un archivo Python normal, no un notebook
from pyspark.sql import DataFrame
from pyspark.sql.functions import col, when, to_date, regexp_replace

def clean_nulls(df: DataFrame, columns: list, fill_value: any = 0) -> DataFrame:
    """
    Limpia valores nulos en columnas específicas.
    
    Args:
        df: DataFrame de entrada
        columns: Lista de columnas a limpiar
        fill_value: Valor para rellenar nulos
    
    Returns:
        DataFrame limpio
    """
    for column in columns:
        df = df.fillna({column: fill_value})
    return df

def standardize_dates(df: DataFrame, date_columns: list, format: str = "yyyy-MM-dd") -> DataFrame:
    """
    Estandariza columnas de fecha.
    """
    for column in date_columns:
        df = df.withColumn(column, to_date(col(column), format))
    return df

def remove_outliers(df: DataFrame, column: str, std_threshold: float = 3.0) -> DataFrame:
    """
    Elimina outliers basados en desviación estándar.
    """
    stats = df.select(
        mean(col(column)).alias("mean"),
        stddev(col(column)).alias("stddev")
    ).collect()[0]
    
    mean_val = stats["mean"]
    stddev_val = stats["stddev"]
    
    return df.filter(
        (col(column) >= mean_val - std_threshold * stddev_val) &
        (col(column) <= mean_val + std_threshold * stddev_val)
    )
```

```python
    # Los parámetros llegan como JSON string
    params = json.loads(sys.argv[1])
else:
    # Valores por defecto para desarrollo
    params = {
        "fecha_proceso": "2024-03-20",
        "modo": "incremental",
        "enviar_alertas": True
    }

fecha = params["fecha_proceso"]
modo = params["modo"]
print(f"🚀 Iniciando ETL para fecha: {fecha}, modo: {modo}")

# Importar módulos
from src.etl.extract import extract_ventas_dia
from src.etl.transform import transform_ventas
from src.etl.load import load_ventas
from src.monitoring.alerts import send_alert

try:
    # Extraer
    df_ventas = extract_ventas_dia(fecha)
    
    # Transformar
    df_transformado = transform_ventas(df_ventas)
    
    # Cargar
    num_registros = load_ventas(df_transformado, modo)
    
    # Alerta si hay anomalías
    if params["enviar_alertas"] and num_registros < 100:
        send_alert(f"⚠️ Pocas ventas en {fecha}: {num_registros} registros")
    
    print(f"✅ Proceso completado. {num_registros} registros procesados.")
    
except Exception as e:
    print(f"❌ Error en ETL: {str(e)}")
    if params["enviar_alertas"]:
        send_alert(f"❌ Fallo ETL {fecha}: {str(e)}")
    raise
```

```python
from pyspark.sql import Row
import random
from datetime import datetime, timedelta

# Generar datos de ventas (500 transacciones)
categorias = ["Electrónica", "Ropa", "Hogar", "Deportes"]
productos = {
    "Electrónica": ["Portátil", "Móvil", "Tablet", "Auriculares"],
    "Ropa": ["Camiseta", "Pantalón", "Zapatos", "Chaqueta"],
    "Hogar": ["Sartén", "Toalla", "Lámpara", "Cojín"],
    "Deportes": ["Pelota", "Raqueta", "Bicicleta", "Pesa"]
}
regiones = ["Norte", "Sur", "Centro", "Este"]

ventas = []
fecha_inicio = datetime(2024, 1, 1)

for i in range(1, 501):
    categoria = random.choice(categorias)
    producto = random.choice(productos[categoria])
    region = random.choice(regiones)
    fecha = fecha_inicio + timedelta(days=random.randint(0, 89))
    cantidad = random.randint(1, 5)
    precio = round(random.uniform(10, 500), 2)
    importe = round(cantidad * precio, 2)
    
    # Introducir algunos datos sucios aleatoriamente
    if random.random() < 0.05:  # 5% de datos con problemas
        if random.random() < 0.5:
            region = None
        else:
            cantidad = -1
    
    ventas.append((
        f"V{i:04d}", fecha.date(), categoria, producto, 
        region, cantidad, precio, importe
    ))

df_final = spark.createDataFrame(ventas, [
    "venta_id", "fecha", "categoria", "producto",
    "region", "cantidad", "precio_unitario", "importe"
])

df_final.write.mode("overwrite").saveAsTable("main.default.ventas_final")

print("✅ Dataset creado: main.default.ventas_final")
print(f"📊 Registros: {df_final.count()}")
display(df_final.limit(10))
```
